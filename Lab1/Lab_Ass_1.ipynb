```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, roc_auc_score, roc_curve,
    confusion_matrix, ConfusionMatrixDisplay
)

sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
heart_disease = fetch_ucirepo(id=45) 
  
# data (as pandas dataframes) 
X = heart_disease.data.features 
y = heart_disease.data.targets 
  
# metadata 
print(heart_disease.metadata) 
  
# variable information 
print(heart_disease.variables) 
```

```python
# Combine X and y into one working dataframe
heart_df = pd.concat([X, y], axis=1)
target_col = y.columns[0]   

print("Shape:", heart_df.shape)
print("Target column detected as:", target_col)
heart_df.head()
```

```python
print(heart_df.dtypes)

print(heart_df.describe().T)

missing_summary = heart_df.isna().sum()
missing_summary = missing_summary[missing_summary > 0]
print("Columns with missing values:\n", missing_summary if not missing_summary.empty else "None found")
```

```python
# Separate handling: median for continuous, mode for categorical/discrete
continuous_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]
categorical_cols = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]

for col in continuous_cols:
    if heart_df[col].isna().any():
        median_val = heart_df[col].median()
        heart_df[col] = heart_df[col].fillna(median_val)
        print(f"Filled {col} with median: {median_val}")

for col in categorical_cols:
    if heart_df[col].isna().any():
        mode_val = heart_df[col].mode(dropna=True)[0]
        heart_df[col] = heart_df[col].fillna(mode_val)
        print(f"Filled {col} with mode: {mode_val}")

assert heart_df.isna().sum().sum() == 0, "Still missing values after imputation"
print("Missing values remaining:", heart_df.isna().sum().sum())
```

```python
# Binarize the target
heart_df["disease_present"] = (heart_df[target_col] > 0).astype(int)

print(heart_df["disease_present"].value_counts(normalize=True).rename("proportion"))
```

```python
# EDA: feature distributions
numeric_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
axes = axes.flatten()

for ax, col in zip(axes, numeric_cols):
    sns.histplot(data=heart_df, x=col, hue="disease_present", kde=True, ax=ax, bins=20, palette="Set2")
    ax.set_title(col)

axes[-1].axis("off")
plt.tight_layout()
plt.show()
```

```python
# EDA: correlation structure
corr_cols = numeric_cols + ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal", "disease_present"]
corr_matrix = heart_df[corr_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Feature correlation matrix")
plt.tight_layout()
plt.show()
```

```python
# logistic regression
feature_cols = [c for c in heart_df.columns if c not in [target_col, "disease_present"]]

X_model = heart_df[feature_cols]
y_model = heart_df["disease_present"]

X_train, X_test, y_train, y_test = train_test_split(
    X_model, y_model, test_size=0.2, random_state=7, stratify=y_model
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

baseline_clf = LogisticRegression(max_iter=1000, random_state=7)
baseline_clf.fit(X_train_scaled, y_train)

y_pred  = baseline_clf.predict(X_test_scaled)
y_proba = baseline_clf.predict_proba(X_test_scaled)[:, 1]
```

```python
# Evaluation
acc = accuracy_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print(f"Accuracy : {acc:.3f}")
print(f"ROC AUC  : {auc:.3f}")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm, display_labels=["No disease", "Disease"]).plot(ax=ax[0], cmap="Blues", colorbar=False)
ax[0].set_title("Confusion Matrix")

fpr, tpr, _ = roc_curve(y_test, y_proba)
ax[1].plot(fpr, tpr, label=f"AUC = {auc:.3f}", color="darkorange")
ax[1].plot([0, 1], [0, 1], linestyle="--", color="gray")
ax[1].set_xlabel("False Positive Rate")
ax[1].set_ylabel("True Positive Rate")
ax[1].set_title("ROC Curve")
ax[1].legend()

plt.tight_layout()
plt.show()
```

```python
import wfdb

record_name = "100"


ecg_record = wfdb.rdrecord(record_name, pn_dir="mitdb")
ecg_annotation = wfdb.rdann(record_name, "atr", pn_dir="mitdb")

print("Signal shape:", ecg_record.p_signal.shape)
print("Sampling frequency:", ecg_record.fs, "Hz")
print("Channel names:", ecg_record.sig_name)
print("Number of annotated beats:", len(ecg_annotation.sample))
```

```python
import numpy as np
import pandas as pd

# MLII is channel 0 — the standard lead used for beat detection in most MIT-BIH studies
ecg_signal = ecg_record.p_signal[:, 0]
fs = ecg_record.fs

ann_samples = ecg_annotation.sample
ann_symbols = ecg_annotation.symbol

annotation_df = pd.DataFrame({"sample_index": ann_samples, "beat_type": ann_symbols})
annotation_df["time_sec"] = annotation_df["sample_index"] / fs

print("Beat type counts:\n", annotation_df["beat_type"].value_counts())
print("\nFirst few annotations:")
print(annotation_df.head())
print(f"\nRecording duration: {len(ecg_signal)/fs:.1f} seconds")
```

# 4. Textual data demonstration

```python

import re
from collections import Counter

# Small set of common stopwords — no external NLTK download needed
STOPWORDS = set("""
a an the is are was were be been being of to in on at for with and or but
if then so this that these those it its as by from not no nor has have had
""".split())

clinical_note = """
Patient is a 58-year-old male presenting with intermittent chest pain
radiating to the left arm and jaw, occurring mostly during physical
exertion and relieved by rest. He reports associated shortness of breath
and mild dizziness. No history of fever, cough, or recent trauma.
Patient denies palpitations but mentions occasional fatigue over the past
two weeks. Resting ECG shows nonspecific ST-T wave changes. Blood pressure
recorded at 145/90 mmHg. Advised further cardiac workup including stress
test, lipid profile, and troponin levels. Patient counseled on lifestyle
modification and started on low-dose aspirin pending further evaluation.
"""

def tokenize(text):
    text = text.lower()
    tokens = re.findall(r"[a-z]+", text)
    return tokens

raw_tokens = tokenize(clinical_note)
content_tokens = [t for t in raw_tokens if t not in STOPWORDS]

print("Total tokens:", len(raw_tokens))
print("Content tokens (stopwords removed):", len(content_tokens))
```

```python
# Word frequency and key term extraction
term_freq = Counter(content_tokens)
top_terms = term_freq.most_common(10)

print("Top 10 clinically relevant terms:")
for term, count in top_terms:
    print(f"  {term:15s} {count}")

# Simple bar chart of top terms
terms, counts = zip(*top_terms)
plt.figure(figsize=(9, 4))
plt.bar(terms, counts, color="cornflowerblue", edgecolor="black")
plt.title("Top terms in clinical note (frequency)")
plt.xticks(rotation=45, ha="right")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

# 5. Image data: synthetic chest X-ray, load, display, resize

```python
from PIL import Image, ImageDraw, ImageFilter
import numpy as np

def generate_synthetic_xray(width=300, height=350, seed=1):
   
    rng = np.random.default_rng(seed)
    cx, cy = width // 2, height // 2

    # Base tissue brightness with a soft radial vignette (darker at corners, like real film)
    yy, xx = np.mgrid[0:height, 0:width]
    dist_from_center = np.sqrt((xx - cx) ** 2 + (yy - cy) ** 2)
    base = 200 - (dist_from_center / dist_from_center.max()) * 60

    # Lung fields: darker, slightly uneven ellipses
    left_lung = (((xx - (cx - 65)) ** 2) / (60 ** 2) + ((yy - (cy - 10)) ** 2) / (105 ** 2)) < 1
    right_lung = (((xx - (cx + 65)) ** 2) / (60 ** 2) + ((yy - (cy - 10)) ** 2) / (105 ** 2)) < 1
    base[left_lung | right_lung] -= 95

    # Diaphragm curve cutting off the bottom of each lung field
    diaphragm = yy > (cy + 95 - 0.002 * (xx - cx) ** 2)
    base[diaphragm & (left_lung | right_lung)] += 70

    pixels = np.clip(base, 0, 255).astype(np.uint8)
    img = Image.fromarray(pixels, mode="L")
    img = img.filter(ImageFilter.GaussianBlur(radius=2))

    # Draw faint rib lines and spine on top using PIL's drawing API
    draw = ImageDraw.Draw(img)
    for i, rib_y in enumerate(range(cy - 90, cy + 60, 22)):
        draw.arc([cx - 130, rib_y - 15, cx - 10, rib_y + 60], start=200, end=340, fill=140, width=2)
        draw.arc([cx + 10, rib_y - 15, cx + 130, rib_y + 60], start=200, end=340, fill=140, width=2)

    draw.line([(cx, cy - 120), (cx, cy + 110)], fill=190, width=6)  # spine

    img = img.filter(ImageFilter.GaussianBlur(radius=1))

    # Light film grain
    grain = rng.normal(0, 5, size=(height, width))
    final_arr = np.clip(np.array(img).astype(float) + grain, 0, 255).astype(np.uint8)
    return Image.fromarray(final_arr, mode="L")

xray_img = generate_synthetic_xray()
xray_img.save("synthetic_chest_xray.png")

plt.figure(figsize=(5, 5.5))
plt.imshow(xray_img, cmap="gray")
plt.title("Synthetic chest X-ray (original)")
plt.axis("off")
plt.show()
```

```python
resized_xray = xray_img.resize((128, 128))

fig, axes = plt.subplots(1, 2, figsize=(9, 4.5))
axes[0].imshow(xray_img, cmap="gray")
axes[0].set_title(f"Original {xray_img.size}")
axes[0].axis("off")

axes[1].imshow(resized_xray, cmap="gray")
axes[1].set_title(f"Resized {resized_xray.size}")
axes[1].axis("off")
plt.tight_layout()
plt.show()

pixel_array = np.array(xray_img)
image_metadata = {
    "format": xray_img.format,
    "mode": xray_img.mode,
    "size_px": xray_img.size,
    "min_intensity": int(pixel_array.min()),
    "max_intensity": int(pixel_array.max()),
    "mean_intensity": round(float(pixel_array.mean()), 2),
    "std_intensity": round(float(pixel_array.std()), 2),
}

print("Image metadata:")
for k, v in image_metadata.items():
    print(f"  {k}: {v}")
```

## 6. Integration Discussion — Toward a Fused Multimodal Model

In this lab, we work with three types of datasets:

1. **Tabular clinical data**
2. **ECG signals**
3. **Free-text clinical notes**

Each type of data gives us a different view of a patient's cardiovascular (heart-related) risk. Some information may overlap, but each dataset also provides information that the others may not have.

A **fused model** can combine information from all three datasets to make better predictions.



## 1. Tabular + ECG Fusion

The current **Logistic Regression** model uses information from a single patient visit.

For example, the tabular dataset contains:

- Resting blood pressure (`trestbps`)
- Cholesterol (`chol`)
- Maximum heart rate achieved (`thalach`)
- ST depression (`oldpeak`)
- Other clinical measurements

However, these values mainly describe the patient's condition at one particular time.

An **ECG recording** can provide information about how the heart behaves over a period of time.

We can extract useful features from a continuous ECG recording, such as:

- **Mean R-R interval** — the average time between two heartbeats.
- **SDNN** — a measure of heart-rate variability (HRV).
- **Beat-type irregularity** — for example, the percentage of heartbeats that are not normal (`N` beats).
- **ST-segment deviation over time** — changes in the ST segment during the ECG recording.

These ECG features can be converted into numerical values and added as new columns to `heart_df` before training the model.

## 2. Text Notes as Context

Clinical notes contain symptoms such as **"chest pain radiating to the left arm"** or **"shortness of breath on exertion"**, which may not be available in the tabular data or ECG signal.

Using simple **NLP**, we can extract features such as:

- Whether important symptom words like **"pain," "dizziness," or "fatigue"** are present.
- How often important medical terms appear.

These features can be added as new columns to the dataset, giving the model **early information about the patient's symptoms**, sometimes even before lab results or ECG recordings are available.


## 3. Practical Data Fusion Approach

To combine all three types of data, we need to make sure that the information belongs to the correct patient and the correct time.

A combined dataset could look like this:

Below is a hypothetical example of what a fused table *would* look like once ECG features are engineered and merged in (this was not implemented in this lab, since ECG and tabular data here come from different, unrelated patients).

| Patient ID | Tabular Features | ECG Features | Text Features | Target |
|------------|------------------|---------------|---------------|--------|
| 001        | Age, BP, Chol    | SDNN, RR      | Chest pain    | 1      |
| 002        | Age, BP, Chol    | SDNN, RR      | Dizziness     | 0      |

The fusion process requires three main steps.

### Step 1: Common Patient ID

We need a **shared patient ID** that connects:

- Tabular clinical records
- ECG recordings
- Clinical notes

This allows us to know which ECG and notes belong to which patient.

### Step 2: Feature Engineering

Each type of data must be converted into a fixed-size numerical format.

For example:

- Tabular data → BP, cholesterol, age, etc.
- ECG → mean R-R interval, SDNN, irregularity count, ST deviation, etc.
- Text → symptom word counts or presence/absence values

All these features can then be placed in the same row.

### Step 3: Train a Model

After combining the features, we can train a single classifier using all the information.

For example:

- Tabular features
- ECG features
- Text features
- Combined feature table
- Logistic Regression
- Heart disease prediction
